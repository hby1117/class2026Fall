# Audio Processing

## 1. Waveform

```python
# Import packages
import librosa
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load audio
audio, sr = librosa.load("aeiou.wav")
```

```python
# Visualize audio
librosa.display.waveshow(audio, sr=sr)
```

## 2. Spectrogram

```python
# Pre-emphasis
pre_audio = librosa.effects.preemphasis(audio)
```

```python
# Short-time Fourier Transform (STFT)
n_fft = 2048
win_length = int(sr * 0.005)
hop_length = int(sr * 0.001)

stft_result = librosa.stft(
    pre_audio,
    n_fft=n_fft,
    win_length=win_length,
    hop_length=hop_length
)
```

```python
# Amplitude -> dB
magnitude = np.abs(stft_result)
spec_db = librosa.amplitude_to_db(magnitude, ref=np.max(magnitude))
```

```python
# Visualize spectrogram
plt.figure(figsize=(13, 3))

librosa.display.specshow(
    spec_db,
    sr=sr,
    hop_length=hop_length,
    x_axis="time",
    y_axis="linear",
    cmap="gray_r",
    vmin=-40,
    vmax=0
)

plt.ylim(0, 5000)
plt.xlim(0, len(audio) / sr)
plt.grid(axis="y")
plt.xlabel("Time (s)")
plt.ylabel("Frequency (Hz)")
plt.show()
```
