## 2. External Package

### Numpy
```python
import numpy as np

# Numpy array
a = np.array([1, 2, 3])
```
```python
# Shape
np.shape(a)
```
```python
# Element-wise calculation
a + 2
a - 2
a * 2
a / 2
a ** 2
```
```python
# Basic functions
np.sum(a)
np.mean(a)
np.min(a)
np.max(a)
```
